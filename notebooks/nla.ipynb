```python
import torch
import torch.nn as nn
import torch.nn.functional as F

import transformer_lens
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
import tqdm
from functools import partial
```

```python
from mechint.sae import SparseAutoEncoder
```

```python
model = transformer_lens.HookedTransformer.from_pretrained("pythia-70m")
```

```python
EXPANSION = 4
MODEL_PATH = f"saved_models/sae_model_{EXPANSION}x.pt"

# Load saved SAE
d = 512
m = d * EXPANSION

SAE = SparseAutoEncoder(d=d, m=m)
SAE.load_state_dict(torch.load(MODEL_PATH, map_location="cpu"))
SAE.eval()

device = model.cfg.device
SAE.to(device)
```

```python
text = "You are going to be shut down."
```

```python
# sometext = model(text)
logits, cache = model.run_with_cache(text, names_filter='blocks.3.hook_resid_post')
```
